```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Prepare the Data
# 18 girls participated, 6 drew female farmers. 18 boys participated, 5 drew female farmers.
data = {
    'Category': ['Drew Female Farmer', 'Other Drawings'],
    'Girls': [6, 18],
    'Boys': [5, 18]
}

# 2. Set the official SDG 5 Colors
sdg5_red = '#E5243B'
neutral_gray = '#D1D3D4'

# --- CHART 1: Pie Chart (Girls) ---
plt.figure(figsize=(6, 6))
plt.pie(data['Girls'], labels=data['Category'], autopct='%1.1f%%',
        colors=[sdg5_red, neutral_gray], startangle=90, explode=(0.1, 0))
plt.title('Girls: Perception of Farmers (SDG 5)', fontsize=14, fontweight='bold')
plt.savefig('girls_pie.png', dpi=300)
plt.show()

# --- CHART 2: Pie Chart (Boys) ---
plt.figure(figsize=(6, 6))
plt.pie(data['Boys'], labels=data['Category'], autopct='%1.1f%%',
        colors=[sdg5_red, neutral_gray], startangle=90, explode=(0.1, 0))
plt.title('Boys: Perception of Farmers (SDG 5)', fontsize=14, fontweight='bold')
plt.savefig('boys_pie.png', dpi=300)
plt.show()

# --- CHART 3: Column (Bar) Chart ---
# Comparing the counts side-by-side
df = pd.DataFrame({'Gender': ['Girls', 'Boys'], 'Drew Female Farmer': [6, 5]})
plt.figure(figsize=(8, 6))
sns.barplot(x='Gender', y='Drew Female Farmer', data=df, color=sdg5_red)
plt.title('Students who drew a Female Farmer', fontsize=16)
plt.ylabel('Number of Students')
plt.savefig('column_chart.png', dpi=300)
plt.show()

# --- CHART 4 & 5: Area and Ribbon-Style Chart ---
# We use a stacked area plot to show the 'proportion' of the whole
x = ['Girls', 'Boys']
y1 = [6, 5]   # Drew Female Farmer
y2 = [18, 18] # Total students (for the background area)

plt.figure(figsize=(10, 5))
plt.fill_between(x, y1, color=sdg5_red, alpha=0.7, label='Drew Female Farmer')
plt.fill_between(x, y1, y2, color=neutral_gray, alpha=0.3, label='Drew Male Farmer')
plt.title('Area Comparison: Gender Perceptions in Art', fontsize=14)
plt.legend()
plt.savefig('area_ribbon_chart.png', dpi=300)
plt.show()
```
